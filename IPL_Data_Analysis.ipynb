**IPL - Data Analysis**

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("matches.csv")
```

📊**DataSet Overveiw:**

```python
df.shape
```

```python
df.info()
```

```python
df.head()
```

🧹**Data Cleaning:**

```python
df.isnull().sum()
```

```python
df['city'] = df['city'].fillna("Unkwown")
```

```python
df['player_of_match'] = df['player_of_match'].fillna("No Award")
```

```python
df['winner'] = df['winner'].fillna("No Result")
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['umpire1','umpire2'],errors = 'ignore')
```

```python
df.head()
```

🔍**Matches Per Season:**

```python
df['season'].value_counts().sort_index()
```

🏆**Most Sucessfull Team:**

```python
df['winner'].value_counts()
```

How many matches ended with NO RESULT:

```python
(df['winner'] == "No result").sum()
```

Toss DEcisons - Does winning toss matter???

```python
toss_match = (df['toss_winner'] == df['winner']).sum()
toss_match
```

```python
total_match = df.shape[0]
round((toss_match/total_match)*100)
```
